```python
from collections import deque
```

```python
q = deque()

for i in range(10):
    q.append(i)
```

```python
print(q)
print(len(q))
```

```python
for i, v in enumerate(q):
    q[i] += 1
    
print(q)
print(len(q))
```

```python
for i in range(5):
    q.pop()
    
print(q)
print(len(q))
```
